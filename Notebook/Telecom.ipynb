```python
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3  
import pandas as pd

df = pd.read_csv('../DATA/cleaned_churn_data.csv')
print(df.head(10))
print(df.info())
print(df.describe())
print(df.shape)
print(df.columns)

print(df.isnull().sum())
print(df.duplicated().sum())
print(df.drop_duplicates())
print(df.shape)
print(df.columns)
print(df.head)

df['Number vmail messages'] = df['Number vmail messages'].fillna(0)
print(df['Number vmail messages'].isnull().sum())

international_plan = df['International plan'].value_counts()
print(international_plan)


Highest_day_charge = df.sort_values(by = 'Total day charge' , ascending = False)
print("Highest day charge:", Highest_day_charge[['State', "Total day charge"]].head(10))

df.to_csv

conn = sqlite3.connect('../DATA/cleaned_churn_data.db')
df.to_sql('churn_data',conn, if_exists='replace',index=False)

print(pd.read_sql_query("SELECT * FROM churn_data LIMIT 10" , conn))
print(pd.read_sql_query(
    'SELECT "Total day charge", ROUND(SUM("Total day charge"), 2) AS TOTALDAYCHARGE '
    'FROM churn_data '
    'GROUP BY "Total day charge" '
    'ORDER BY TOTALDAYCHARGE DESC',
    conn
))
print(pd.read_sql_query(
    'SELECT "Total day minutes", ROUND(SUM("Total day minutes"), 2) AS TotalDayMinutes '
    'FROM churn_data '
    'GROUP BY "Total day minutes" '
    'ORDER BY TotalDayMinutes DESC' , 
    conn
))
print(pd.read_sql_query(
    'SELECT "State", ROUND(SUM("Total day charge"), 2) AS TotalDayCharge '
    'FROM churn_data '
    'GROUP BY "State" '
    'ORDER BY TotalDayCharge DESC',
    conn
  ))

plt.style.use('ggplot')

Total_day_charge = df.groupby('State')['Total day charge'].sum()
print(Total_day_charge)
plt.plot(kind = 'bar')
plt.title('Total Day Charge by State')
plt.xlabel('State')
plt.ylabel("Total day charge")
plt.show()

Total_day_minutes = df.groupby('State')['Total day minutes'].sum()
print(Total_day_minutes)
plt.plot(kind = 'bar')
plt.title('Total Day Minutes by State')
plt.xlabel('State')
plt.ylabel("Total day minutes")
plt.show()

State_day_charge = df.groupby('State')['Total day charge'].sum()
print(State_day_charge)
plt.plot(kind = 'bar')
plt.title("Total day charge  by state")
plt.xlabel('State')
plt.ylabel("Total day charge")
plt.show()

correlation = df[['Area code', 'Total day charge', 'Total day minutes']].corr()
print(correlation)
print(correlation)
plt.figure(figsize=(10, 6))
sns.heatmap(correlation, annot = True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()

Total_day_charge.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Total day charge distribution by state')
plt.ylabel('')
plt.show()
Total_day_minutes.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Total day minutes distribution by state')
plt.ylabel('')
plt.show()
```
